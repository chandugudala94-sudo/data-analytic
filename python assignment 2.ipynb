```python
# Create age list
age_list = [24, 25, 26, 27, 28]

# Create name list
name_list = ["Arun", "Bala", "Chandru", "Deepak", "Kiran"]

print(age_list)
print(name_list)
```

```python
# a. Append Yazhini
name_list.append("Yazhini")
print(name_list)

# b. Insert 30 at index 2
age_list.insert(2, 30)
print(age_list)

# c. Remove Yazhini
name_list.remove("Yazhini")
print(name_list)

# d. Pop the last element
age_list.pop()
print(age_list)

# e. Extend age_list
age_list.extend([29, 30, 26])
print(age_list)

# f. Sort in descending order
age_list.sort(reverse=True)
print(age_list)

# g. Find maximum, minimum and sum
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python
# Create dictionary
student_marks = {
    "Arun": 75,
    "Bala": 68,
    "Chandru": 90,
    "Deepak": 82,
    "Kiran": 70
}

print(student_marks)

# Access a specific student's mark
print("Chandru's mark:", student_marks["Chandru"])

# Add Janani
student_marks["Janani"] = 80

# Update an older student's mark
student_marks["Bala"] = 82

print(student_marks)

# Print all keys
print("Keys:", student_marks.keys())

# Print all values
print("Values:", student_marks.values())

# Print key-value pairs
print("Items:", student_marks.items())
```
